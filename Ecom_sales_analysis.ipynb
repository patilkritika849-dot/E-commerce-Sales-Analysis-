```python
import pandas as pd
import matplotlib.pyplot as plt

# Data
data = {
    'Product': ['Mobile', 'Laptop', 'Headphone', 'Mobile', 'Laptop', 'Watch'],
    'Sales': [40000, 70000, 2000, 38000, 75000, 5000],
    'Profit': [4000, 7000, 500, 3500, 8000, 1000],
    'Month': ['Jan', 'Jan', 'Jan', 'Feb', 'Feb', 'Feb']
}
df = pd.DataFrame(data)

print(df)
print("\nwhich is the most selled products?")
print(df.groupby('Product')['Sales'].sum())

# to showing a graph in colab
df.groupby('Product')['Sales'].sum().plot(kind='bar', color='orange')
plt.title('Sales Project ')
plt.show()
```
